# California Student Access on Cal-ICOR

Explore enrollment, run analysis, and see charts — all in the browser.

```python
import pandas as pd
import matplotlib.pyplot as plt

enrollment = pd.DataFrame({
    "campus": ["CCC", "CSU", "UC", "Private"],
    "students_k": [1800, 460, 290, 120],
})
enrollment
```

```python
fig, ax = plt.subplots(figsize=(8, 3.6))
colors = ["#1B4F72", "#1a56db", "#0e7490", "#F4A261"]
bars = ax.bar(enrollment["campus"], enrollment["students_k"], color=colors, width=0.62)
ax.set_ylabel("Students (thousands)")
ax.set_title("Approximate enrollment by segment")
ax.spines[["top", "right"]].set_visible(False)
for bar, value in zip(bars, enrollment["students_k"]):
    ax.text(bar.get_x() + bar.get_width() / 2, value + 40, str(value), ha="center", fontsize=10)
plt.tight_layout()
plt.show()
```

```python
share = enrollment.assign(pct=lambda d: (100 * d["students_k"] / d["students_k"].sum()).round(1))
share.sort_values("pct", ascending=False)
```
